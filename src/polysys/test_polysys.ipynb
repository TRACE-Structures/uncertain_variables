```python
from uncertain_variables.polysys import ChebyshevTPolynomials
from uncertain_variables.polysys import ChebyshevUPolynomials
from uncertain_variables.polysys import HermitePolynomials
from uncertain_variables.polysys import JacobiPolynomials
from uncertain_variables.polysys import LaguerrePolynomials
from uncertain_variables.polysys import LegendrePolynomials

import numpy as np
```

```python
from scipy.special import chebyt

type(chebyt)
```

```python
x = np.linspace(-1, 1, 5)

chebyt = ChebyshevTPolynomials(orthonormal=False)
chebyt_orth = ChebyshevTPolynomials(orthonormal=True)

chebyt_values = chebyt.evaluate(deg=3, x=x)
chebyt_orth_values = chebyt_orth.evaluate(deg=3, x=x)

print("ChebyshevTPolynomials (non-orthonormal) values:")
print(chebyt_values)
print("ChebyshevTPolynomials (orthonormal) values:")
print(chebyt_orth_values)

chebyt_derivative_values = chebyt.evaluate(deg=3, x=x, derivative=True)
chebyt_orth_derivative_values = chebyt_orth.evaluate(deg=3, x=x, derivative=True)

print("ChebyshevTPolynomials (non-orthonormal) derivative values:")
print(chebyt_derivative_values)
print("ChebyshevTPolynomials (orthonormal) derivative values:")
print(chebyt_orth_derivative_values)
```
